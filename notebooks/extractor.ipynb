### Caso de prueba: Extraer datos de una factura

```python
# Pipeline de extraccion de datos de facturas
pdf_paths = "data/invoices"
print(f"Processing path {pdf_paths}...")
file_list = invoice_extractor.enumerate_pdf_files(pdf_paths)
print(f"Processing {len(file_list)} files...")
print(f"Processing first file: {file_list[0]}...")
invoice_extractor.process_pdf_files(file_list)
```

```python
# Lectura de datos de facturas
print("Reading data...")
processed_json = [file for file in os.listdir(pdf_paths) if file.endswith(".json")]
df_invoices = pd.DataFrame()
for file in processed_json:    
    temp = pd.read_json(f"{pdf_paths}/{file}")
    temp.insert(0, "file", file)
    df_invoices = pd.concat([df_invoices, temp])
df_invoices.head(5)
```

```python
pdf_paths = "data/medical-studies"
print(f"Processing path {pdf_paths}...")
file_list = medical_features_extractor.enumerate_pdf_files(pdf_paths)
print(f"Processing {len(file_list)} files...")
print(f"Processing first file: {file_list[0]}...")
medical_features_extractor.process_pdf_files(file_list)
```

```python
# Reading data
print("Reading data...")
processed_json = [file for file in os.listdir(pdf_paths) if file.endswith(".json")]
df_medical_features = pd.DataFrame()
for file in processed_json:    
    with open(f"{pdf_paths}/{file}", "r") as f:
        json_data = json.load(f)  # Load the JSON data
        if isinstance(json_data, dict):  # Wrap dictionary in a list
            json_data = [json_data]
        temp = pd.DataFrame(json_data)  # Convert to DataFrame
        temp.insert(0, "file", file)  # Add the filename as a new column
        df_medical_features = pd.concat([df_medical_features, temp], ignore_index=True)  # Concatenate with main DataFrame
df_medical_features.head(5)
```

Este traajo utiliza el framework de LLamaIndex para procesar de forma estructurada estudios medicos. 
Se utiliza el modelo pago gpt-4o de OpenAI para realizar la extraccion de cada feature, un schema Pydantic para validar la estructura de los datos obtenidos como asi tambien para dar las directivas al modelo de como interpretar cada campo y no menos importante un PromptTemplate para indicarle el comportamiento deseado.
